#### Instrucción 3: 
<p style='text-align: justify;'>Crea una nueva variable <b>x</b> donde grabes toda el dataframe excepto el target y una nueva variable <b>y</b> donde guardes la columna de target. Una vez que termines lo anterior utilizar la paquetería de "<b>model_selection</b>" para separar nuestro conjunto de datos en 70% para entrenamiento y 30% para la prueba.</p>

```python
x = df.drop('target',axis=1)
print(x.shape)
y = df['target']
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.7,random_state=100,stratify=y)

print(f'Tamanio de x entrenamiento: {x_train.shape}')
print(f'Tamanio de x de prueba    : {x_test.shape}')
print(f'Tamanio de y de entrenamiento: {y_train.shape}')
print(f'Tamanio de y de prueba       : {y_test.shape}')
```

#### Instrucción 4: 
<p style='text-align: justify;'>Utilizando el split de datos, entrena y compara un modelo de clasificación basado en una "<b>SVM con kernel linear</b>". Utiliza tus datos para construir un reporte de clasficiación y grafica la matriz de confusión resultante usando un mapa de calor.</p>

```python
from sklearn import svm
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
# Cramos maquina de soporte vectorial para clasificacion
maquina = svm.SVC(kernel='linear')

#La entrenamos con los datos
maquina = maquina.fit(x_train.values,y_train.values)

#Hacemos predicciones con la maquina de soporte vectorail
y_train_predict = maquina.predict(x_train.values)
y_test_predict = maquina.predict(x_test.values)
y_predict = maquina.predict(x.values)

#Creamos matrices de confuncion
mat_train = confusion_matrix(y_true = y_train.values,y_pred=y_train_predict)
#print(mat_train)
mat_test = confusion_matrix(y_true = y_test.values,y_pred=y_test_predict)
#print(mat_test)
mat = confusion_matrix(y_true = y.values,y_pred=y_predict)
#print(mat)

#Mostramos matrices con mapas de calor
fig,ax = plt.subplots(1,3,figsize=(10,3))

sns.heatmap(mat_train,annot=True,cmap="summer", linewidth=0.5,ax=ax[0],cbar=False,fmt='g')
ax[0].set_title("Mapa de calor de entrenamiento \n de MSV con kernel lineal")
sns.heatmap(mat_test,annot=True,cmap="summer", linewidth=0.5, ax=ax[1],cbar=False)
ax[1].set_title("de conjunto de prueba")
sns.heatmap(mat,annot=True,cmap="summer",linewidth=0.5,ax=ax[2],fmt='g')
ax[2].set_title(" y de conjunto total")
plt.show()

#Dados los bugs de las versiones de sns, imprimiremos los mapas de forma numérica
for i,j,k in zip(mat_train,mat_test,mat):
    print(f'{i}-{j}-{k}')

#Imprimimos reportes de clasificacion
print('-'*100)
print('Reporte de clasificación de datos de entrenamiento:')
print(classification_report(y_true=y_train,y_pred=y_train_predict))

print('-'*100)
print('Reporte de clasificación de datos de prueba:')
print(classification_report(y_true=y_test,y_pred=y_test_predict))

print('-'*100)
print('Reporte de clasificación de datos totales:')
print(classification_report(y_true=y,y_pred=y_predict))
```

#### Instrucción 5: 
<p style='text-align: justify;'>Reproduce las siguientes celdas para encontrar el vector de soporte y el ranking para identificar las diez caracteristicas más importantes que impactan al modelo de clasificación. </p>

```python
from sklearn.feature_selection import RFE
selector = RFE(estimator = maquina, n_features_to_select=10, step = 1)
selector = selector.fit(x_train, y_train)
print(selector.support_)
print(selector.ranking_)
```

```python
n = int(selector.support_.size)

for i in range(n):
    if selector.support_[i] == True:
        print("Columna " + str(i))
```
