#### Instrucción 3: 
<p style='text-align: justify;'>En la siguiente sección de nuestra libreta  analizaremos a la variable "<b>Body mass index</b>" (bmi). Crea una variable <b>x</b> donde grabes la columna <b>bmi</b> y una variable <b>y</b> donde almacenemos la información de la columna <b>target</b>. <b>Nota:</b> es muy recomendable a cada una de las variables les des la forma de una columna de datos con ayuda de la instrucción <b>values.reshape(-1, 1)</b>.</p>

```python
x = df['bmi'].values.reshape(-1,1)
y = pd.DataFrame(diabetes.target).values.reshape(-1,1)

for i,j in zip(x[:10],y[:10]):
    print(f'x:{i}, - y:{j}')
```

#### Instrucción 4: 
<p style='text-align: justify;'>Divide el conjunto de datos en entrenamiento y prueba. Asigna el 80 porciento de los datos para el entrenamiento y el resto para la prueba.</p>

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=100)

print(f'Tamanio de conjunto de entrenamiento: {x_train.shape}')
print(f'Tamanio de conjunto de pruebas:       {x_test.shape}')
print(f'353 + 89 = {len(x)}')
print('-'*50)
print(f'Tamanio de Y de entrenamiento: {y_train.shape}')
print(f'Tamanio de Y de pruebas:       {y_test.shape}')
```

#### Instrucción 5: 
<p style='text-align: justify;'>Con ayuda de la paquetería de sklearn construye un modelo de <b>regresión polinomial de grado 2</b> que se ajuste a nuestros datos de entrenamiento y utiliza el modelo para obtener los dos siguientes conjuntos de respuestas: "<b>y_train_predict</b>" y "<b>y_test_predict</b>".</p>

<p style='text-align: justify;'>Posterior a construir el modelo utilizando la función <b>plt.scatter</b> construye una grafica donde se muestren el conjuto de entrenamiento en color azul, el conjunto de prueba en naranja y una línea recta roja donde se muestre la predicción devuelta por nuestro modelo.</p>

Por último obten la metrica r2_score para el modelo obtenido

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

#Creamos el objeto que transformará a nuestros conjutnos
poly = PolynomialFeatures(2)

# Transformamos los datos de entrenamiento y de prueba
x_train_transform = poly.fit_transform(x_train)
x_test_transform = poly.fit_transform(x_test)

# Creamos modelo vacio
model = LinearRegression()

#Entrenamos el modelo con datos transformados
model = model.fit(x_train_transform,y_train)

#Vemos las predicciones sobre los datos de entrenamiento y los de prueba
y_train_predict = model.predict(x_train_transform)
y_test_predict = model.predict(x_test_transform)


#Creamos grafica de modelo
x_model = np.linspace(np.min(x),np.max(x),100)
y_model = model.predict(poly.fit_transform(x_model.reshape(-1,1)))

#Grafica 
plt.plot(figsize=(10,8))
plt.scatter(x_train,y_train,label='Entrenamiento')
plt.scatter(x_test,y_test,label = 'Prueba')
plt.plot(x_model,y_model,'r--',label = 'Prediccion \nModelo')
plt.legend(loc='lower right')
plt.title('Grafico de dispersion')
plt.show()

# R2 score
r2_regresion_polinomio2_entrenamiento = r2_score(y_true = y_train,y_pred=y_train_predict)
r2_regresion_polinomio2_prueba = r2_score(y_true=y_test,y_pred=y_test_predict)
print(f'r2_score entrenamiento : {r2_regresion_polinomio2_entrenamiento}')
print(f'r2_score prueba:       : {r2_regresion_polinomio2_prueba}')

```

```python
from sklearn.svm import SVR
from sklearn.metrics import r2_score
from sklearn.preprocessing import MinMaxScaler

#Creamos escalador 
scaler = MinMaxScaler()

#Lo entrenamos con los datos de entrenamiento
scaler = scaler.fit(x)

#Ajustamos 
x_train_standar = scaler.transform(x_train).reshape(-1,1)
x_test_standar = scaler.transform(x_test).reshape(-1,1)



#Creamos modelo vacío
maquina = SVR(kernel='linear',C=100)
maquina = maquina.fit(x_train_standar,y_train)

#Creamos predicciones
y_train_predict = maquina.predict(x_train_standar)
y_test_predict = maquina.predict(x_test_standar)

# Datos para graficar modelo
x_model = np.linspace(np.min(x),np.max(x),100)
y_model = maquina.predict(scaler.transform(x_model.reshape(-1,1)))

plt.plot(figsize=(10,8))
plt.scatter(x_train,y_train,label='Entrenamiento')
plt.scatter(x_test,y_test,label='Prueba')
plt.plot(x_model,y_model,'r--',label='Modelo de \nSVM')
plt.legend()
plt.show()


r2_regresion_svmlineal_entrenamiento = r2_score(y_train,y_train_predict)
r2_regresion_svmlineal_prueba = r2_score(y_test,y_test_predict)
print(f'r2_score de entrenamiento: {r2_regresion_svmlineal_entrenamiento}')
print(f'r2_score de prueba       : {r2_regresion_svmlineal_prueba}')
```

#### Instrucción 8: 
<p style='text-align: justify;'>Con ayuda de la paquetería de sklearn construye un modelo de <b>regresión basado en una maquina de soporte vectorial con un nucleo RBF</b> que se ajuste a nuestros datos de entrenamiento y utiliza el modelo para obtener los dos siguientes conjuntos de respuestas: "<b>y_train_predict</b>" y "<b>y_test_predict</b>".</p>

<p style='text-align: justify;'>Posterior a construir el modelo utilizando la función <b>plt.scatter</b> construye una grafica donde se muestren el conjuto de entrenamiento en color azul, el conjunto de prueba en naranja y una línea recta roja donde se muestre la predicción devuelta por nuestro modelo.</p>

Por último obten la metrica r2_score para el modelo obtenido

```python
from sklearn.svm import SVR
from sklearn.metrics import r2_score
from sklearn.preprocessing import MinMaxScaler


#Creamos escalador 
scaler = MinMaxScaler()

#Lo entrenamos con los datos de entrenamiento
scaler = scaler.fit(x)

#Ajustamos 
x_train_standar = scaler.transform(x_train).reshape(-1,1)
x_test_standar = scaler.transform(x_test).reshape(-1,1)



#Creamos modelo vacío
maquina = SVR(kernel='rbf',C=100)
maquina = maquina.fit(x_train_standar,y_train)

#Creamos predicciones
y_train_predict = maquina.predict(x_train_standar)
y_test_predict = maquina.predict(x_test_standar)

# Datos para graficar modelo
x_model = np.linspace(np.min(x),np.max(x),100)
y_model = maquina.predict(scaler.transform(x_model.reshape(-1,1)))

plt.plot(figsize=(10,8))
plt.scatter(x_train,y_train,label='Entrenamiento')
plt.scatter(x_test,y_test,label='Prueba')
plt.plot(x_model,y_model,'r--',label='Modelo de \nSVM')
plt.legend()
plt.show()

r2_regresion_svmrbf_entrenamiento = r2_score(y_train,y_train_predict)
r2_regresion_svmrbf_prueba = r2_score(y_test,y_test_predict)
print(f'r2_score de entrenamiento: {r2_regresion_svmrbf_entrenamiento}')
print(f'r2_score de prueba       : {r2_regresion_svmrbf_prueba}')
```

#### Instrucción 9: 
<p style='text-align: justify;'>Con ayuda de la paquetería de sklearn construye un modelo de <b>árbol de regresión de profundidad 5</b> que se ajuste a nuestros datos de entrenamiento y utiliza el modelo para obtener los dos siguientes conjuntos de respuestas: "<b>y_train_predict</b>" y "<b>y_test_predict</b>".</p>

<p style='text-align: justify;'>Posterior a construir el modelo utilizando la función <b>plt.scatter</b> construye una grafica donde se muestren el conjuto de entrenamiento en color azul, el conjunto de prueba en naranja y una línea recta roja donde se muestre la predicción devuelta por nuestro modelo.</p>

Por último obten la metrica r2_score para el modelo obtenido

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
from sklearn.preprocessing import MinMaxScaler



#Creamos escalador 
scaler = MinMaxScaler()

#Lo entrenamos con los datos de entrenamiento
scaler = scaler.fit(x)

#Ajustamos 
x_train_standar = scaler.transform(x_train).reshape(-1,1)
x_test_standar = scaler.transform(x_test).reshape(-1,1)



#Creamos modelo vacío
arbol = DecisionTreeRegressor(max_depth=5)
arbol = arbol.fit(x_train_standar,y_train)

#Creamos predicciones
y_train_predict = arbol.predict(x_train_standar)
y_test_predict = arbol.predict(x_test_standar)

# Datos para graficar modelo
x_model = np.linspace(np.min(x),np.max(x),100)
y_model = arbol.predict(scaler.transform(x_model.reshape(-1,1)))

plt.plot(figsize=(10,8))
plt.scatter(x_train,y_train,label='Entrenamiento')
plt.scatter(x_test,y_test,label='Prueba')
plt.plot(x_model,y_model,'r--',label='Modelo de \nSVM',linewidth=2)
plt.legend()
plt.show()

r2_arbol_entrenamiento = r2_score(y_train,y_train_predict)
r2_arbol_prueba = r2_score(y_test,y_test_predict)
print(f'r2_score de entrenamiento: {r2_arbol_entrenamiento}')
print(f'r2_score de prueba       : {r2_arbol_prueba}')
```
